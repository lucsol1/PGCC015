```python
import pandas as pd
import random
```

```python
def signal(u):
    return 1 if u >= 0 else 0
```

```python
def classificar(x,weights):
    x_k = [-1] + list(x)
    u = sum(x_k[i] * weights[i] for i in range(len(weights)))
    y = signal(u)

    classe = "C2" if y == 1 else "C1"
    return y, classe
```

```python
#Perceptron
def perceptron(X, d, h=0.1, verbose=True):
    n_inputs = len(X[0])

    
    weights = [
        random.uniform(0, 1) 
        for _ in range(n_inputs + 1)
    ]

    if verbose:
        print(f"Epoca {epoch} ---- Pesos: {weights}")
    epoch = 0

    while True:
        error = False

        for k in range(len(X)):
            #entrada do bias -1 + inputs originais
            x_k = [-1] + list(X[k])

            u = sum(
                x_k[i] * weights[i] 
                for i in range(len(weights))
            )

            y = signal(u)

            if d[k] != y:
                #wi <- wi-1 + h.(d(k) - y).x(k)
                for i in range(len(weights)):
                    weights[i] += h * (d[k] - y) * x_k[i]

                error = True 

        epoch += 1
        if not error:
            break
    return weights
```

```python
with open('data/treinamento2.txt', 'r', encoding='utf-8') as file:
    read_file = file.read().split()

input_train = []
output_train = []

for i in range(0, len(read_file), 4):
    grupo = read_file[i:i+4]
    if len(grupo) == 4:
        input_train.append([float(x) for x in grupo[:3]])
        output_train.append(int(grupo[3]))

print("Inputs:", input_train)
print("Outputs:", output_train)
```

```python
with open('data/teste2.txt', 'r', encoding='utf-8') as file:
    read_file = file.read().split()

input_test = []
ouput_test = []

for i in range(0, len(read_file), 4):
    grupo = read_file[i:i+4]
    if len(grupo) == 4:
        input_test.append([float(x) for x in grupo[:3]])
        ouput_test.append(int(grupo[3]))

print("Inputs:", input_test)
print("Outputs:", ouput_test)
```

```python
resultados_treinamentos = []

for treino in range(1, 6):
    random.seed(treino)
    pesos_iniciais = [random.uniform(0, 1) for _ in range(len(input_train[0]) + 1)]
    print(f"Treinamento {treino} - pesos iniciais: {pesos_iniciais}")

    random.seed(treino)
    pesos_finais = perceptron(input_train, output_train, h=0.01, verbose=False)

    resultados_treinamentos.append({
        "treino": treino,
        "pesos_iniciais": pesos_iniciais,
        "pesos_finais": pesos_finais
    })

for r in resultados_treinamentos:
    print(f"\nTreinamento {r['treino']}")
    print(f"Pesos iniciais: {r['pesos_iniciais']}")
    print(f"Pesos finais: {r['pesos_finais']}")
```
