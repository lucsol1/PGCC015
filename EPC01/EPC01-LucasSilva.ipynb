```python
import pandas as pd
import random
```

## 1. Preparação e Treinamento

### 1.1 Função Perceptron e Funções Complementares

```python
def signal(u):
    return 1 if u >= 0 else -1
```

```python
def predict(x,weights):
    x_k = [-1] + list(x)
    u = sum(x_k[i] * weights[i] for i in range(len(weights)))
    y = signal(u)

    classe = "C2" if y == 1 else "C1"
    return y, classe
```

```python
def perceptron(X, d, learning_rate=0.1, verbose=True):
    n_inputs = len(X[0])

    
    weights = [
        random.uniform(0, 1) 
        for _ in range(n_inputs + 1)
    ]

    
    epoch = 0

    while True:
        error = False

        for k in range(len(X)):
            #entrada do bias
            x_k = [-1] + list(X[k])

            u = sum(
                x_k[i] * weights[i] 
                for i in range(len(weights))
            )

            y = signal(u)

            if d[k] != y:
                #wi <- wi-1 + h.(d(k) - y).x(k)
                for i in range(len(weights)):
                    weights[i] += learning_rate * (d[k] - y) * x_k[i]

                error = True 

        epoch += 1

        if verbose:
                print(f"Epoca {epoch} ---- Pesos: {weights}")

        if not error:
            break
    return weights, epoch
```

### 1.2 Leitura dos arquivos

```python
with open('data/treinamento2.txt', 'r', encoding='utf-8') as file:
    read_file = file.read().split()

input_train = []
output_train = []

for i in range(0, len(read_file), 4):
    grupo = read_file[i:i+4]
    if len(grupo) == 4:
        input_train.append([float(x) for x in grupo[:3]])
        # fiz isso porquê o enunciado diz -1 e 1 na função sinal, porém o dataset está em [0, 1], 
        y = int(grupo[3])
        output_train.append(1 if y == 1 else -1) 

print("Inputs:", input_train)
print("Outputs:", output_train)
```

```python
with open('data/teste2.txt', 'r', encoding='utf-8') as file:
    read_file = file.read().split()

input_test = []
ouput_test = []

for i in range(0, len(read_file), 4):
    grupo = read_file[i:i+4]
    if len(grupo) == 4:
        input_test.append([float(x) for x in grupo[:3]])
        y = int(grupo[3])
        ouput_test.append(1 if y == 1 else -1) 

print("Inputs:", input_test)
print("Outputs:", ouput_test)
```

## 2. Relatório

#### **Atividade 1 -** Execute 5 treinamentos para a rede Perceptron inicializando o vetor de pesos em cada treinamento com valores aleatórios entre zero e um. Se for o caso, reinicie o gerador de números aleatórios em cada treinamento de tal forma que os elementos do vetor de pesos iniciais não sejam os mesmos.

```python
resultados_treinamentos = []

for treino in range(1, 6):
    random.seed(treino) 
    pesos_iniciais = [random.uniform(0, 1) for _ in range(len(input_train[0]) + 1)]

    # só pra garantir os mesmos pesos inicias
    random.seed(treino)
    pesos_finais, num_epocas = perceptron(input_train, output_train, learning_rate=0.01, verbose=False)

    resultados_treinamentos.append({
        "treino": treino,
        "pesos_iniciais": pesos_iniciais,
        "pesos_finais": pesos_finais,
        "num_epocas": num_epocas
    })

for r in resultados_treinamentos:
    print(f"\nTreinamento {r['treino']} (epocas: {r['num_epocas']})")
    print(f"  Pesos iniciais: {r['pesos_iniciais']}")
    print(f"  Pesos finais:   {r['pesos_finais']}")
```

#### **Atividade 2 -** Registre os resultados dos 5 treinamentos na tabela abaixo:

```python
linhas_tabela1 = []
for r in resultados_treinamentos:
    linhas_tabela1.append({
        "Treinamento": f"T{r['treino']}",
        "w0_inicial": round(r["pesos_iniciais"][0], 4),
        "w1_inicial": round(r["pesos_iniciais"][1], 4),
        "w2_inicial": round(r["pesos_iniciais"][2], 4),
        "w3_inicial": round(r["pesos_iniciais"][3], 4),
        "w0_final": round(r["pesos_finais"][0], 4),
        "w1_final": round(r["pesos_finais"][1], 4),
        "w2_final": round(r["pesos_finais"][2], 4),
        "w3_final": round(r["pesos_finais"][3], 4),
        "N. Epocas": r["num_epocas"]
    })
    

tabela1 = pd.DataFrame(linhas_tabela1)
tabela1
```

#### **Atividade 3 -** Após o treinamento do Perceptron aplique o mesmo na classificação automática das seguintes amostras de óleo, indicando na tabela abaixo os resultados das saídas (Classes) referentes aos cinco processos de treinamento realizados no item 1.

```python
linhas_tabela2 = []
for idx, x in enumerate(input_test, start=1):
    linha = {"Amostra": idx, "x1": x[0], "x2": x[1], "x3": x[2]}
    for r in resultados_treinamentos:
        # usei para pegar a classe e não o valor(y) bruto [0,1]
        _, classe = predict(x, r["pesos_finais"])   
        linha[f"Classe (T{r['treino']})"] = classe      
    linhas_tabela2.append(linha)

tabela2 = pd.DataFrame(linhas_tabela2)
tabela2
```

#### **Atividade 4 -** Explique por que o número de épocas de treinamento varia a cada vez que executamos o treinamento do Perceptron.

O Perceptron pode encontrar mais de uma reta que separe as classes, vide que a configuração final do vetor de pesos é fortemente dependente dos valores inciais que foram aleatoriamente alocados

#### **Atividade 5 -** Qual a principal limitação do Perceptron quando aplicado em problemas de classificação de padrões.

Precisa ser possível traçar uma reta no espaço afim de separar as classes, se o problema não for lineramente separável, o perceptron não conseguirá resolver. Além disso, caso o problema seja linearmente separável, mas a fronteira de decisão for pequena, o perceptron pode demorar muito para convergir ou sequer conseguir traçar essa reta.
