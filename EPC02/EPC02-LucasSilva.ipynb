### Função Adaline e Funções Complementares

```python
#erro quadrático médio
def eqm(X, d, weights):
    p = len(X)
    erro = 0
    for k in range(p):
        x_k = [-1] + list(X[k])
        u = sum(x_k[i] * weights[i] for i in range(len(weights)))
        erro += (d[k] - u) ** 2
    return erro / p
```

```python
def signal(u):
    return 1 if u >= 0 else 0
```

```python
# Rede Adaline
def adaline(X, d, h=0.0025, erro_tolerado=1e-2, pesos_iniciais=None):
    n_inputs = len(X[0])
 
    if pesos_iniciais is None:
        weights = [random.uniform(0, 1) for _ in range(n_inputs + 1)]
    else:
        weights = list(pesos_iniciais)
 
    pesos_iniciais_usados = list(weights)
 
    EQM_ant = float("inf")
    EQM_atual = eqm(X, d, weights)
    epoca = 0
    historico_eqm = [EQM_atual]
 
    while abs(EQM_atual - EQM_ant) > erro_tolerado:
        EQM_ant = EQM_atual
 
        for k in range(len(X)):
            x_k = [-1] + list(X[k])
            u = sum(x_k[i] * weights[i] for i in range(len(weights)))
            for i in range(len(weights)):
                weights[i] += h * (d[k] - u) * x_k[i]
 
        EQM_atual = eqm(X, d, weights)
        epoca += 1
        historico_eqm.append(EQM_atual)
 
    return weights, epoca, historico_eqm, pesos_iniciais_usados
```

```python
#acuracia e desvio padrao
def testar_rede(X_teste, d_teste, weights):
    acertos = 0
    for k in range(len(X_teste)):
        x_k = [-1] + list(X_teste[k])
        u = sum(x_k[i] * weights[i] for i in range(len(weights)))
        saida = signal(u)
        if saida == d_teste[k]:
            acertos += 1
    return acertos / len(X_teste)
```

### Treinamento

```python
def rodar_experimento(nome_dataset, X_train, d_train, X_test, d_test,
                       h=0.0025, erro_tolerado=1e-2):
    resultados = []
 
    for treino in range(1, 4):

        random.seed(treino)  
 
        pesos_finais, num_epocas, historico_eqm, pesos_iniciais = adaline(
            X_train, d_train, h=h, erro_tolerado=erro_tolerado
        )
 
        acuracia = testar_rede(X_test, d_test, pesos_finais)
 
        resultados.append({
            "dataset": nome_dataset,
            "treino": treino,
            "pesos_iniciais": pesos_iniciais,
            "pesos_finais": pesos_finais,
            "num_epocas": num_epocas,
            "historico_eqm": historico_eqm,
            "acuracia": acuracia,
        })
 
        print(f"[{nome_dataset}] Treino {treino} | épocas: {num_epocas} "
              f"| EQM final: {historico_eqm[-1]:.6f} | acurácia: {acuracia:.4f}")
 
    acuracias = [r["acuracia"] for r in resultados]
    desvio_padrao = statistics.stdev(acuracias) if len(acuracias) > 1 else 0.0
    media = statistics.mean(acuracias)
 
    print(f"\n[{nome_dataset}] Acurácia média: {media:.4f} | "
          f"Desvio padrão: {desvio_padrao:.4f}\n")
 
    return resultados, media, desvio_padrao
```

```python
def criar_tabela_pesos_epocas(resultados):
    linhas = []
    for r in resultados:
        linhas.append({
            "Dataset": r["dataset"],
            "Treino": r["treino"],
            "Épocas": r["num_epocas"],
            "Pesos iniciais": [round(w, 4) for w in r["pesos_iniciais"]],
            "Pesos finais": [round(w, 4) for w in r["pesos_finais"]],
        })
    return pd.DataFrame(linhas)
 
 
def criar_tabela_acuracia(resultados, media, desvio_padrao, nome_dataset):
    linhas = [{"Dataset": nome_dataset, "Treino": r["treino"],
               "Acurácia": round(r["acuracia"], 4)} for r in resultados]
    linhas.append({"Dataset": nome_dataset, "Treino": "Média",
                    "Acurácia": round(media, 4)})
    linhas.append({"Dataset": nome_dataset, "Treino": "Desvio padrão",
                    "Acurácia": round(desvio_padrao, 4)})
    return pd.DataFrame(linhas)
```

```python
def plotar_eqm(resultados_separaveis, resultados_sobrepostos):
    fig, axs = plt.subplots(2, 3, figsize=(15, 8))
    fig.suptitle("EQM por época — Adaline (3 treinamentos)")
 
    for r in resultados_separaveis:
        idx = r["treino"] - 1
        axs[0, idx].plot(r["historico_eqm"])
        axs[0, idx].set_title(f"Separáveis - Treino {r['treino']}")
        axs[0, idx].set_xlabel("Época")
        axs[0, idx].set_ylabel("EQM")
 
    for r in resultados_sobrepostos:
        idx = r["treino"] - 1
        axs[1, idx].plot(r["historico_eqm"], color="orange")
        axs[1, idx].set_title(f"Sobrepostos - Treino {r['treino']}")
        axs[1, idx].set_xlabel("Época")
        axs[1, idx].set_ylabel("EQM")
 
    plt.tight_layout()
    plt.show()
```

### Respostas

```python
"""
1 - Execute três treinamentos para a rede ADALINE inicializando o vetor de pesos em cada
treinamento com valores aleatórios entre zero e um. Se for o caso, reinicie o gerador de
números aleatórios em cada treinamento de tal forma que os elementos do vetor de pesos
iniciais não sejam os mesmos. Utilize taxa de aprendizado h = 0.0025 e precisão e = 10-2.
Você pode tentar outros valores para estes parâmetros para tentar melhorar o aprendizado da
rede.
"""

resultados_separaveis, media_sep, dp_sep = rodar_experimento(
    "Separáveis",
    input_train_separaveis, output_train_separaveis,
    input_test_separaveis, ouput_test_separaveis,
    h=0.0025, erro_tolerado=1e-2
)

resultados_sobrepostos, media_sob, dp_sob = rodar_experimento(
    "Sobrepostos",
    input_train_sobrepostos, output_train_sobrepostos,
    input_test_sobrepostos, output_test_sobrepostos,
    h=0.0025, erro_tolerado=1e-2
)
```

```python
tabela_pesos = pd.concat([
    criar_tabela_pesos_epocas(resultados_separaveis),
    criar_tabela_pesos_epocas(resultados_sobrepostos),
], ignore_index=True)

tabela_acuracia = pd.concat([
    criar_tabela_acuracia(resultados_separaveis, media_sep, dp_sep, "Separáveis"),
    criar_tabela_acuracia(resultados_sobrepostos, media_sob, dp_sob, "Sobrepostos"),
], ignore_index=True)
```

```python
def salvar_tabela_como_imagem(df, nome_arquivo, titulo=""):
    fig, ax = plt.subplots(figsize=(len(df.columns) * 2, len(df) * 0.5 + 1))
    ax.axis("off")
    if titulo:
        ax.set_title(titulo, fontweight="bold")
    tabela = ax.table(cellText=df.values, colLabels=df.columns,
                       cellLoc="center", loc="center")
    tabela.auto_set_font_size(False)
    tabela.set_fontsize(10)
    tabela.scale(1, 1.2)
    plt.tight_layout()
    plt.savefig(nome_arquivo, dpi=250, bbox_inches="tight")
    plt.show()
```
