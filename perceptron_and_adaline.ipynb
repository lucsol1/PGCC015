```python
import pandas as pd
import numpy as np
import random
```

```python
def signal(u):
    return 1 if u >= 0 else -1
```

```python
def test(x, weights):
    x_k = [-1] + list(x)
    u = sum(x_k[i] * weights[i] for i in range(len(weights)))
    y = signal(u)

    classe = "Classe 1" if y == 1 else "Classe 2"
    return y, classe
```

```python
#Perceptron
def perceptron(X, d, h=0.1):
    n_inputs = len(X[0])

    weights = [
        random.uniform(0, 1) 
        for _ in range(n_inputs + 1)
    ]

    epoch = 0

    while True:
        error = False

        for k in range(len(X)):
            #entrada do bias -1 + inputs originais
            x_k = [-1] + list(X[k])

            u = sum(
                x_k[i] * weights[i] 
                for i in range(len(weights))
            )

            y = signal(u)

            if d[k] != y:
                #wi <- wi-1 + h.(d(k) - y).x(k)
                for i in range(len(weights)):
                    weights[i] += h * (d[k] - y) * x_k[i]

                error = True 

        epoch += 1

        if not error:
            break
    return weights
```

```python
X = [
    [-1, -1],
    [-1,  1],
    [ 1, -1],
    [ 1,  1],
]
d = [-1, -1, -1, 1]

w = perceptron(X, d, h=0.1)
print("Pesos finais:", w)

entrada = input("\nDigite o padrão x separado por vírgula (ex: 1,-1): ")
x = [float(valor) for valor in entrada.split(",")]

y, classe = test(x, w)
print(f"Padrão {x} -> y = {y} -> {classe}")
```

```python
def eqm(X, d, weights):
    """
    EQM <- 0
    Para cada par de treinamento p {x(k),d(k)} Faça
        u <- x(k)^T * w
        EQM <- EQM + (d(k) - u)^2
    Fim_para
    EQM <- EQM / p
    """
    p = len(X)
    eqm = 0

    for k in range(p):
        x_k = [-1] + list(X[k])
        u = sum(x_k[i] * weights[i] for i in range(len(weights)))
        eqm += (d[k] - u) ** 2

    eqm = eqm / p
    return eqm

def adaline(X, d, h=0.01, erro_tolerado=1e-6):
    n_inputs = len(X[0])

    weights = [
        random.uniform(-1, 1)
        for _ in range(n_inputs + 1)
    ]

    EQM_ant = float("inf")
    EQM_atual = 1

    epoca = 0

    while abs(EQM_atual - EQM_ant) > erro_tolerado:
        EQM_ant = EQM_atual

        for k in range(len(X)):
            x_k = [-1] + list(X[k])

            # u <- x(k)^T * w
            u = sum(x_k[i] * weights[i] for i in range(len(weights)))

            # w <- w + h . (d(k) - u) . x(k)
            for i in range(len(weights)):
                weights[i] += h * (d[k] - u) * x_k[i]

        EQM_atual = eqm(X, d, weights)

        epoca += 1
        print(f"Época {epoca:4d} | EQM = {EQM_atual:.6f}")

    return weights


if __name__ == "__main__":
    X = [
        [-1, -1],
        [-1,  1],
        [ 1, -1],
        [ 1,  1],
    ]
    d = [-1, -1, -1, 1]

    w = adaline(X, d, h=0.01, erro_tolerado=1e-6)
    print("\nPesos finais:", w)

    print("\nTeste:")
    for x in X:
        y, classe = test(x, w)
        print(f"Entrada: {x} -> y = {y} -> {classe}")
```
